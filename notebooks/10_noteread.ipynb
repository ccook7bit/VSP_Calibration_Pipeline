```python
import argparse
import glob
import os
import pandas as pd

def open_dat_file(file):
    try:
        with open(file, 'r', encoding='utf-8') as f:
            data = f.readlines()
    except UnicodeDecodeError:
        with open(file, 'r', encoding='latin-1') as f:
            data = f.readlines()
    
    print(f"Opened file: {file}, number of lines read: {len(data)}")
    print(f"First few lines of the raw data:\n{''.join(data[:10])}")
    return [line.strip().split() for line in data]

def get_dat_files(path):
    if os.path.isdir(path):
        os.chdir(path)
        dat_files = glob.glob("*.dat")
    else:
        dat_files = [path]
    print(f"Found .dat files: {dat_files}")
    return dat_files

def main(path):
    dat_files = get_dat_files(path)
    all_data = []

    for dat_file in dat_files:
        data = open_dat_file(dat_file)
        all_data.extend(data)

    if not all_data:
        print("No data found in .dat files.")
    else:
        # Assuming the data has columns and we need columns 1, 2, and 3 (0-indexed)
        # Adjust the column indices as per your data structure
        columns_to_extract = [0, 1, 2]  # Example indices
        extracted_data = []
        
        for line in all_data:
            if len(line) >= max(columns_to_extract) + 1:
                extracted_data.append([line[i] for i in columns_to_extract])
            else:
                print(f"Skipping line due to insufficient columns: {line}")

        # Create DataFrame
        df = pd.DataFrame(extracted_data, columns=['EXPTIME', 'IMAGENAME', 'RA'])  # Adjust column names
        
        print(df.head())
```
