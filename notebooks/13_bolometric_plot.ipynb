```python
def lineAt(x, d):
    return d
def oneDFit(a,b,x):
    return a*x+b
def psuedoBolo(data, kronCut, kronDist, kronBand, bitFlagsTF, bitFlags, max):
    print(kronCut, kronDist, kronBand, bitFlags, max)
    counter = 0
    RotseData = []
    PanData = []
    binFlags = []
    if(bitFlagsTF):
        for n in range(max):
            binFlags.append(np.binary_repr(int(data[n][5]), width=8))
            if(kronCut):
                if float(data[n][4]) > 5.0 and float(data[n][4]) < 25.0 and binFlags[n] != bitFlags and float(data[n][kronBand]) - float(data[n][kronBand + 1]) > kronDist or float(data[n][kronBand]) - float(data[n][kronBand + 1]) < -1*kronDist:
                    x = float(data[n][4])  
                    gflux = pow(10, (float(data[n][6])+48.6)/-2.5)
                    rflux = pow(10, (float(data[n][8])+48.6)/-2.5)
                    iflux = pow(10, (float(data[n][10])+48.6)/-2.5)
                    zflux = pow(10, (float(data[n][12])+48.6)/-2.5)
                    yflux = pow(10, (float(data[n][14])+48.6)/-2.5)
                    totalFlux = (gflux*(7.058-5.454) + rflux*(5.454-4.347) + iflux*(4.347-3.68) + zflux*(3.68-3.26) +yflux*(3.26-3))/((7.058-3))
                    logpart = math.log(totalFlux/3631e-23, 10)
                    y = -2.5*logpart
                    RotseData.append(x)
                    PanData.append(y)
                    counter = counter + 1
            else:
                if float(data[n][4]) > 5.0 and float(data[n][4]) < 25.0 and binFlags[n] != bitFlags: 
                    x = float(data[n][4])   
                    gflux = pow(10, (float(data[n][6])+48.6)/-2.5)
                    rflux = pow(10, (float(data[n][8])+48.6)/-2.5)
                    iflux = pow(10, (float(data[n][10])+48.6)/-2.5)
                    zflux = pow(10, (float(data[n][12])+48.6)/-2.5)
                    yflux = pow(10, (float(data[n][14])+48.6)/-2.5)
                    totalFlux = (gflux*(7.058-5.454) + rflux*(5.454-4.347) + iflux*(4.347-3.68) + zflux*(3.68-3.26) +yflux*(3.26-3))/((7.058-3))
                    logpart = math.log(totalFlux/3631e-23, 10)
                    y = -2.5*logpart
                    RotseData.append(x)
                    PanData.append(y)
                    counter = counter + 1    
    else:
        for n in range(max):
            binFlags.append(np.binary_repr(int(data[n][5]), width=8))
            if(kronCut):
                if float(data[n][4]) > 5.0 and float(data[n][4]) < 25.0 and float(data[n][kronBand]) - float(data[n][kronBand + 1]) > kronDist or float(data[n][kronBand]) - float(data[n][kronBand + 1]) < -1*kronDist:
                    x = float(data[n][4])  
                    gflux = pow(10, (float(data[n][6])+48.6)/-2.5)
                    rflux = pow(10, (float(data[n][8])+48.6)/-2.5)
                    iflux = pow(10, (float(data[n][10])+48.6)/-2.5)
                    zflux = pow(10, (float(data[n][12])+48.6)/-2.5)
                    yflux = pow(10, (float(data[n][14])+48.6)/-2.5)
                    totalFlux = (gflux*(7.058-5.454) + rflux*(5.454-4.347) + iflux*(4.347-3.68) + zflux*(3.68-3.26) +yflux*(3.26-3))/((7.058-3))
                    logpart = math.log(totalFlux/3631e-23, 10)
                    y = -2.5*logpart
                    RotseData.append(x)
                    PanData.append(y)
                    counter = counter + 1
            else:
                if float(data[n][4]) > 5.0 and float(data[n][4]) < 25.0: 
                    x = float(data[n][4])   
                    gflux = pow(10, (float(data[n][6])+48.6)/-2.5)
                    rflux = pow(10, (float(data[n][8])+48.6)/-2.5)
                    iflux = pow(10, (float(data[n][10])+48.6)/-2.5)
                    zflux = pow(10, (float(data[n][12])+48.6)/-2.5)
                    yflux = pow(10, (float(data[n][14])+48.6)/-2.5)
                    totalFlux = (gflux*(7.058-5.454) + rflux*(5.454-4.347) + iflux*(4.347-3.68) + zflux*(3.68-3.26) +yflux*(3.26-3))/((7.058-3))
                    logpart = math.log(totalFlux/3631e-23, 10)
                    y = -2.5*logpart
                    RotseData.append(x)
                    PanData.append(y)
                    counter = counter + 1
    return RotseData, PanData, counter
```
