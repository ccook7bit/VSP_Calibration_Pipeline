```python
import numpy as np
from astropy.io import fits
import pandas as pd

# Assuming ra_list, dec_list, df_magnitude, and df_errors are already defined

def create_fits_file(ra_list, dec_list, df_magnitude, df_errors, output_filename):
    # Ensure we are working with numpy arrays
    ra_list = np.asarray(ra_list).flatten()
    dec_list = np.asarray(dec_list).flatten()

    # Convert dataframes to numpy arrays
    magnitudes = df_magnitude.to_numpy()
    errors = df_errors.to_numpy()
    
    # Number of objects and exposures
    num_objects = len(ra_list)
    num_exposures = magnitudes.shape[1]
    
    # Create primary HDU (header/data unit)
    primary_hdu = fits.PrimaryHDU()
    
    # Create columns for the binary table
    col1 = fits.Column(name='RA', array=ra_list, format='D')
    col2 = fits.Column(name='Dec', array=dec_list, format='D')
    col3 = fits.Column(name='Magnitude', array=magnitudes, format=f'{num_exposures}D')
    col4 = fits.Column(name='Error', array=errors, format=f'{num_exposures}D')
    
    # Create the binary table HDU
    cols = fits.ColDefs([col1, col2, col3, col4])
    hdu = fits.BinTableHDU.from_columns(cols)
    
    # Create HDU list and write to file
    hdul = fits.HDUList([primary_hdu, hdu])
    hdul.writeto(output_filename, overwrite=True)
    
    print(f"FITS file '{output_filename}' created successfully with {num_objects} objects.")
```
